## Architecture contract

```text
LLDP + fabric API + DCIM
          ↓
versioned topology inventory
          ↓
Node labels → Kueue Topology → ResourceFlavor
          ↓
ClusterQueue / LocalQueue → Workload topologyAssignment
          ↓
kube-scheduler → node
          ↓
kubelet Topology Manager → NUMA-local CPU/GPU/NIC
```

Kueue decides **which cluster topology domain**. Topology Manager decides **whether resources inside one node align**. They are separate owners.

### Production decision exercise

```text
Leaf A: 6/8 GPUs occupied; four local GPUs free in 3 minutes
Leaf B: 4/8 GPUs occupied; four GPUs free now
Cross-leaf fallback: 35% lower throughput
```

The decision is not “always pack.” Evaluate:

`completion time = queue delay + training time`

and include fragmentation, priority, deadline, and cost. Required and preferred encode different business promises.

---
## Gang scheduling and TAS are different

```text
Kueue TAS       → which topology domain has capacity for the PodSet?
Gang scheduling → can the required members start as a group?
Cohort          → which ClusterQueues may borrow quota?
Topology Manager→ are CPU/memory/device allocations NUMA-compatible on one node?
```

Kubernetes 1.37 Workload/PodGroup makes native gang scheduling Beta behind the `GenericWorkload` feature gate. It does not discover your fabric or replace TAS.

## Closing rules

1. GPUs are not fungible.
2. Physical topology needs an owned, versioned data contract.
3. Admit the whole distributed workload before consuming GPUs.
4. Cross-node topology and NUMA topology have different owners.
5. Measure useful model progress, not only allocated GPUs.
